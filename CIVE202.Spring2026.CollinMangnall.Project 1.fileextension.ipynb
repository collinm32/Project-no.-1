```python
# First off, we have to insert our packages, and this is the code to do it.
import pandas as pd
# Then, to load our data and make sure we are using the right resource, we copy it exactly from our given resource.
air_data = pd.read_csv('AirQuality_Daily_StudentVersion.csv')
```

```python
# This is to turn our data into a data frame.
air_data = pd.DataFrame(air_data)
```

```python
# This line will make it so the voc data at the end is grouped by sensor names.
voc_grouped = air_data.groupby('sensor.name')

# This will make it so that the grouped data has a statistical summary that gives me the four main statistics.
air_data_voc = voc_grouped.agg(max = ('voc', 'max'),
                                                    min = ('voc', 'min'),
                                                    mean = ('voc', 'mean'),
                                                    median = ('voc', 'median'))

# After getting the main statistics, this will make it so that they are sorting by mean and by the highest top 5 voc.
top5_mean_voc = (
    air_data_voc.sort_values(by = 'mean', ascending = False) #sort_values sorts some column that i need to specify
)
top5_mean_voc.head(5)

# This will do the same thing, but by median this time.
top5_median_voc = (
    air_data_voc.sort_values(by = 'median', ascending = False) #sort_values sorts some column that i need to specify
)
top5_median_voc.head(5)
```

```python
# This will give humidity a function that assigns humidity levels
def assign_humidity(humidity): 
    if humidity < 50: 
        return 'Low' 
    elif 51 <= humidity <= 80: 
        return 'High' 
    elif 80 <= humidity <= 100:
        return 'Very High' # Humidity above 80 is Very High

# Application of the Function
air_data['humidity_level'] = air_data['humidity'].apply(assign_humidity)

# Group by sensor name and humidity level
humidity_by_sensor = (air_data.groupby(['sensor.name', 'humidity_level']).size())
humidity_by_sensor

```

```python
# This will give temperature a function that assigns temperature levels
def assign_temperature(temperature): 
    if temperature < 32: 
        return 'Below Freezing'
    elif 32 <= temperature <= 50:
        return 'Cool'
    elif 50 <= temperature <= 70:
        return 'Warm'
    elif 70 <= temperature <= 150:
        return 'Hot'
        
# Application of the Function
air_data['temperature_level'] = air_data['temperature'].apply(assign_temperature)

# Group by sensor name and humidity level
temperature_by_sensor = (air_data.groupby(['sensor.name', 'temperature_level']).size())
temperature_by_sensor
```

```python
def assign_PM25(PM25): 
    if PM25 < 12: 
        return 'Good'
    elif 12 <= PM25 <= 35.4:
        return 'Moderate'
    elif 35.4 <= PM25 <= 55.4:
        return 'Unhealthy for Sensitive Group'
    elif 55.4 <= PM25 <= 150.4:
        return 'Unhealthy'
    elif 150.4 <= PM25 <= 250.4:
        return 'Very Unhealthy' 
    elif 250.4 <= PM25 <= 1000:
        return 'Hazardous'
air_data['pm2.5_AQI'] = air_data['pm2.5_atm'].apply(assign_PM25)
air_data
pm25_by_sensor = (air_data.groupby(['sensor.name', 'pm2.5_AQI']).size())
pm25_by_sensor
```

```python

def assign_PM10(PM10): 
    if PM10 < 54: 
        return 'Good'
    elif 54 <= PM10 <= 154:
        return 'Moderate'
    elif 154 <= PM10 <= 254:
        return 'Unhealthy for Sensitive Group'
    elif 254 <= PM10 <= 354:
        return 'Unhealthy'
    elif 354 <= PM10 <= 424:
        return 'Very Unhealthy' 
    elif 424 <= PM10 <= 1000:
        return 'Hazardous'
air_data['pm10.0_AQI'] = air_data['pm10.0_atm'].apply(assign_PM10)
pm10_by_sensor = (air_data.groupby(['sensor.name', 'pm10.0_AQI']).size())
pm10_by_sensor
```
